1. Caesar Cipher

```python
# Caesar Cipher Encryption

text = input("Enter plaintext: ")
key = int(input("Enter key (1-25): "))

if key < 1 or key > 25:
    print("Invalid key!")
else:
    encrypted_text = ""

    for char in text:
        if 'A' <= char <= 'Z':
            encrypted_text += chr((ord(char) - ord('A') + key) % 26 + ord('A'))
        elif 'a' <= char <= 'z':
            encrypted_text += chr((ord(char) - ord('a') + key) % 26 + ord('a'))
        else:
            encrypted_text += char

    print("Encrypted text:", encrypted_text)
```

2: Monoalphabetic Substitution Cipher

```python
# Monoalphabetic Substitution Cipher Encryption

text = "HELLO WORLD"
key = "QWERTYUIOPASDFGHJKLZXCVBNM"

encrypted_text = ""

for char in text:
    if 'A' <= char <= 'Z':
        index = ord(char) - ord('A')
        encrypted_text += key[index]
    elif 'a' <= char <= 'z':
        index = ord(char) - ord('a')
        encrypted_text += key[index].lower()
    else:
        encrypted_text += char

print("Original text:", text)
print("Encrypted text:", encrypted_text)
```

3: Playfair Cipher

```python
key = "MONARCHY"
text = "INSTRUMENTS"

alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
key = key.upper().replace("J", "I")
s = ""

for ch in key + alphabet:
    if ch in alphabet and ch not in s:
        s += ch

matrix = [list(s[i:i+5]) for i in range(0, 25, 5)]

print("Keyword:", key)
print("Plaintext:", text)
print("\nPlayfair Matrix:")

for row in matrix:
    print(*row)

text = text.upper().replace("J", "I")
text = "".join(ch for ch in text if ch.isalpha())

if len(text) % 2 != 0:
    text += "X"

encrypted = ""

for i in range(0, len(text), 2):
    a, b = text[i], text[i+1]
    if a == b:
        b = "X"

    r1, c1 = next((r, c) for r in range(5) for c in range(5) if matrix[r][c] == a)
    r2, c2 = next((r, c) for r in range(5) for c in range(5) if matrix[r][c] == b)

    if r1 == r2:
        encrypted += matrix[r1][(c1+1)%5] + matrix[r2][(c2+1)%5]
    elif c1 == c2:
        encrypted += matrix[(r1+1)%5][c1] + matrix[(r2+1)%5][c2]
    else:
        encrypted += matrix[r1][c2] + matrix[r2][c1]

print("\nEncrypted text:", encrypted)
```

5: Affine Caesar Cipher

```python
import math

text = "AFFINE CIPHER"
a, b = 5, 8

print("Plaintext:", text)
print(f"Key values: a = {a}, b = {b}")

if math.gcd(a, 26) != 1:
    print("Invalid key! a must be coprime with 26.")
else:
    encrypted = ""

    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            p = ord(char) - base
            encrypted += chr((a * p + b) % 26 + base)
        else:
            encrypted += char

    print("Encrypted text:", encrypted)
```

6: Affine Cipher Attack

```python
cipher = "BU BU BUBU"
a, b = 3, 15

inverse = -1

for i in range(1, 26):
    if (a * i) % 26 == 1:
        inverse = i
        break

if inverse == -1:
    print("Invalid affine key!")
else:
    plain = ""

    for char in cipher:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            c = ord(char) - base
            plain += chr((inverse * (c - b)) % 26 + base)
        else:
            plain += char

    print("Ciphertext:", cipher)
    print(f"Affine key: a = {a}, b = {b}")
    print("Inverse of a:", inverse)
    print("Decrypted text:", plain)
```

10: Playfair Cipher Encryption.

```python
matrix = [
    ['M', 'F', 'H', 'I', 'K'],
    ['U', 'N', 'O', 'P', 'Q'],
    ['Z', 'V', 'W', 'X', 'Y'],
    ['E', 'L', 'A', 'R', 'G'],
    ['D', 'S', 'T', 'B', 'C']
]

plaintext = "Must see you over Cadogan West. Coming at once."

# Remove spaces and punctuation
text = ''.join(ch.upper() for ch in plaintext if ch.isalpha())
text = text.replace('J', 'I')

# Prepare letter pairs
prepared = ""
i = 0

while i < len(text):
    first = text[i]
    i += 1

    if i < len(text) and text[i] != first:
        second = text[i]
        i += 1
    else:
        second = 'X'

    prepared += first + second

# Find a letter's position in the matrix
def find_position(ch):
    if ch == 'J':
        ch = 'I'

    for r in range(5):
        for c in range(5):
            if matrix[r][c] == ch:
                return r, c

# Encrypt each pair
encrypted = ""

for i in range(0, len(prepared), 2):
    r1, c1 = find_position(prepared[i])
    r2, c2 = find_position(prepared[i + 1])

    if r1 == r2:
        encrypted += matrix[r1][(c1 + 1) % 5]
        encrypted += matrix[r2][(c2 + 1) % 5]

    elif c1 == c2:
        encrypted += matrix[(r1 + 1) % 5][c1]
        encrypted += matrix[(r2 + 1) % 5][c2]

    else:
        encrypted += matrix[r1][c2]
        encrypted += matrix[r2][c1]

print("Plaintext:", plaintext)
print("Prepared text:", prepared)
print("Encrypted text:", encrypted)
```

11: Number of Possible Playfair Cipher Keys.

```python
import math

# Calculate 25!
total = math.factorial(25)

# Effective arrangements: 25! / 4!
effective = total // math.factorial(4)

print("Total possible Playfair keys (25!):")
print(total)

print("\nEffective keys (25! / 4!):")
print(effective)

print(f"\nApproximate total key space: 2^{math.log2(total):.2f}")
print(f"Approximate effective key space: 2^{math.log2(effective):.2f}")
```

12: Hill Cipher Encryption and Decryption.

```python
import string

plaintext = "meet me at the usual place at ten rather than eight oclock"

K = [[9, 4],
     [5, 7]]

inverse = [[5, 12],
           [11, 25]]

# Remove spaces and convert to uppercase
text = ''.join(ch.upper() for ch in plaintext if ch.isalpha())

if len(text) % 2 != 0:
    text += 'X'

# Hill Cipher Encryption
encrypted = ""

for i in range(0, len(text), 2):
    a = ord(text[i]) - ord('A')
    b = ord(text[i + 1]) - ord('A')

    c1 = (K[0][0] * a + K[0][1] * b) % 26
    c2 = (K[1][0] * a + K[1][1] * b) % 26

    encrypted += chr(c1 + ord('A'))
    encrypted += chr(c2 + ord('A'))

# Hill Cipher Decryption
decrypted = ""

for i in range(0, len(encrypted), 2):
    a = ord(encrypted[i]) - ord('A')
    b = ord(encrypted[i + 1]) - ord('A')

    c1 = (inverse[0][0] * a + inverse[0][1] * b) % 26
    c2 = (inverse[1][0] * a + inverse[1][1] * b) % 26

    decrypted += chr(c1 + ord('A'))
    decrypted += chr(c2 + ord('A'))

print("Plaintext:", text)
print("Encrypted text:", encrypted)
print("Decrypted text:", decrypted)
```

13: Hill Cipher Known-Plaintext Attack.

```python
import math

def mod26(x):
    return (x % 26 + 26) % 26

P = [[7, 4],
     [11, 15]]

C = [[9, 4],
     [5, 7]]

# Calculate determinant of P
det = mod26(P[0][0] * P[1][1] - P[0][1] * P[1][0])

# Find modular inverse of determinant
invDet = -1

for i in range(26):
    if mod26(det * i) == 1:
        invDet = i
        break

if invDet == -1:
    print("Plaintext matrix is not invertible modulo 26.")
else:
    # Calculate inverse of P modulo 26
    Pinv = [
        [mod26(P[1][1] * invDet), mod26(-P[0][1] * invDet)],
        [mod26(-P[1][0] * invDet), mod26(P[0][0] * invDet)]
    ]

    # Recover key matrix K = C * inverse(P) modulo 26
    K = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            for k in range(2):
                K[i][j] += C[i][k] * Pinv[k][j]

            K[i][j] = mod26(K[i][j])

    print("Recovered Key Matrix:")
    for row in K:
        print(*row)
```

14: One-Time Pad Using Vigenère Encryption.

```python
def mod26(x):
    return x % 26

plaintext = "SEND MORE MONEY"

target = "CASH NOT NEEDED"

key = [9, 0, 1, 7, 23, 15, 21, 14, 11, 11, 2, 8, 9]

print("Plaintext:", plaintext)

# Encrypt while preserving spaces
encrypted = ""
j = 0

for ch in plaintext:
    if ch.isalpha():
        p = ord(ch.upper()) - ord('A')
        encrypted += chr(mod26(p + key[j]) + ord('A'))
        j += 1
    else:
        encrypted += ch

print("Encrypted text:", encrypted)

# Generate key stream for target plaintext
print(f'\nKey stream for target plaintext "{target}":')

j = 0
for ch in plaintext:
    if ch.isalpha():
        p = ord(ch.upper()) - ord('A')
        c = mod26(p + key[j])

        t = ord(target[j].upper()) - ord('A')

        print(mod26(c - t), end=" ")
        j += 1

print()
```

15: Frequency Analysis Attack on an Additive Cipher.

```python
frequency = [
    8.167, 1.492, 2.782, 4.253, 12.702,
    2.228, 2.015, 6.094, 6.966, 0.153,
    0.772, 4.025, 2.406, 6.749, 7.507,
    1.929, 0.095, 5.987, 6.327, 9.056,
    2.758, 0.978, 2.360, 0.150, 1.974, 0.074
]

cipher = "WKH TXLFN EURZQ IRA MXPSV RYHU WKH ODCB GRJ"

scores = {}

print("Ciphertext:", cipher)
print("\nTop 10 candidate decryptions:\n")

# Calculate English-frequency score for each shift
for shift in range(26):
    plain = ""
    score = 0.0

    for ch in cipher:
        if ch.isalpha():
            c = ord(ch.upper()) - ord('A')
            p = (c - shift + 26) % 26
            plain += chr(p + ord('A'))
            score += frequency[p]
        else:
            plain += ch

    scores[shift] = (score, plain)

# Display the top 10 candidates
ranked = sorted(scores.items(), key=lambda item: item[1][0], reverse=True)

for shift, (score, plain) in ranked[:10]:
    print(f"Shift {shift:2d} | Score {score:.3f} | {plain}")
```
